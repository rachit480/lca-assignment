```python
def check_right_triangle():
    L = []
    for i in range(3):
        side = int(input("enter side lengths: "))
        L.append(side)
    c = max(L)
    L.remove(c)
    a = L[0]
    b = L[1]
    sum = a**2 + b**2
    if sum == (c**2):
        print("right angled triangle")
    elif sum != (c**2):
        print("not right angled triangle")
check_right_triangle()
```
