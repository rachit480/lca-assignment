```python
# to create a list and perform add, update and delete operations on it 
L = [1,2,3,4]
print("Original list:", L)
L.append(5) #add single element
print("After adding element:", L) 
L.extend([6,7]) #add multiple elements
print("After adding elements:", L)  
L[2] = 9
print("After updating:", L)
L.pop(2) #delete by index
print("After deleting:", L)
L.remove(7) #delete by element
print("After deleting:", L)
# To create a tuple
T = tuple(L)
print("Tuple:", T)
print("tuples are immutable")
# to create dictionary and perform add, update and delete operations on it
D = {1:"Pranit", 2:"Ishan", 3:"Shourya"}
print("Original Dictionary:", D)
D[4] = "Adbhut"
print("After adding element, dictionary is:", D)
D[4] = "Philip"
print("After updating:", D)
del(D[2])
print("After deleting:", D)
```
